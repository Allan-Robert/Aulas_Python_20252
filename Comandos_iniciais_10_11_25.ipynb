```python
## for loops com tamanho definido
### for indexador in intervalo:
l=[2,3,5,7,9,12,15]
# criar uma soma
soma=0
for i in range(len(l)):
  soma=soma+l[i]
print(soma)

### Maneira mais python
soma=0
for indice,valor in enumerate(l):
  soma=soma+valor
print(f'A média deste dados é {soma/(indice+1)}')

```

```python
## Criar uma função com for imbutido
### Um produtório só pra aceitar numeros inteiros
def produtorio(lista):
  if (type(lista)=='int'):
    prod=1
    for i,v in enumerate(lista):
      prod=prod*v
    return prod

produtorio(l)
```

```python
### criar uma função que calcule a série de fibonacci par n valores [1,1,2,3,5,8,13,21,34,...]

def fibon(n):
  fib=[1,1] # dois primeiros números
  for i in range(n-2): # para criar somente n-2 dois elementos dá série (2 primeiros definidos)
    fib.append(fib[-1]+fib[-2]) # append acrescenta um elemento a direita
  return fib

fibon(10)
```
