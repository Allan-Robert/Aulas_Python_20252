```python
## Fatiamento intervalos de uma lista tupla ou string
l=[2,4,3,5,6,7,8,3,5]
l[2:] # da posição 2 em diante
l[:5] # posicoes anteriores a 5
l[-4:] # do -4 em diante
l[:-4] # do começo até o anterior ao -4
l[2:-2] # da posição 2 ate anterior ao -2
```

```python
# Operadores booleanos Ou (or) E (and) Não (not)
A=3
B=2
C=2
### Ou basta que uma setença seja verdade para retornar True
print(A==B or A==C or B==C)
### E todas as sentenças tem que ser verdadeiras para retorna True
print(A==B and A==C and B==C)
### not é a negação de uma sentença
print(not A==B) # True
print(not B==C) # False
```
