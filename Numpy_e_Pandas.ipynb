## Falando do NumPy
O NumPy (Numerical Python) é a biblioteca fundamental para a computação científica em Python.

Características:
* Objeto ndarray (N-dimensional Array): É a estrutura básica do NumPy. É um grade (array) homogêneo e multidimensional mais eficiente em termos de memória e velocidade do que as listas nativas do Python para operações numéricas.
* Velocidade: As operações no NumPy são implementadas em C, o que as torna mais rápidas do que operações em Python puro, em especialmente para grande volume de dados.
* Funções Matemáticas Abrangentes:Oferece uma vasta coleção de funções para operações matemáticas elementares como trigonometria, logaritmos, etc. Ferramentas com métodos de Álgebra linear, além de estatísticas descritivas e distribuições de probabilidade.  


```python
###
import numpy as np
l=[2,1,4,5,6,9,12,22]
### Funções estatísticas e matemáticas
np.sin(l)
np.cos(l)
np.max(l)
np.mean(l)
np.std(l)
np.var(l)
np.median(l)

am=np.random.normal(10,2,20)
type(am)
```

```python
## indexação em numpy é semelhante as lista
am[4:]
am[-1]
```

## Pandas

O Pandas é a biblioteca mais utilizada em Python para manipulação e análise de dados. Ela oferece um estruturas de dados com uma gama de funções para dados como: leitura, limpeza, filtragem, agrupamento, agregação, transformações e descritivas.

Estruturas Chaves:

*Series (Série): Um array rotulado, unidimensional, similar a uma coluna em uma planilha ou um vetor (construído sobre o NumPy).

*DataFrame (Tabela de Dados): A estrutura principal. É uma tabela bidimensional, onde cada coluna é um objeto Series e todas compartilham o mesmo índice (rótulo) de linha.

```python
import pandas as pd

aluno=pd.read_csv("/content/drive/MyDrive/Colab Notebooks/Alunos.csv",encoding='latin1',sep=';')
aluno.head(2)
```

```python
aluno.info() # informações gerais importantes
```

```python
## Como fazer referência as colunas
aluno['Curso'] # referencia direta de uma coluna
aluno[['Curso','Sexo']]# mais de uma coluna
aluno.iloc[:,[1,3,5]] # citar por numeração a coluna (Quando se tem muitas colunas)
```

```python
## 1 Estatística e 2 Ciências Atuariais
def mcurso(num):
  if num==1:
    return 'Estatística'
  else: return 'Ciências Atuariais'

aluno['Curso']=aluno['Curso'].apply(mcurso)
aluno.head(3)
```

```python
## função lambda
aluno['Sexo']=aluno['Sexo'].apply(lambda x: 'Masculino' if x==1 else 'Feminino')
aluno.head(3)
```

```python
## uso do cut 1 publica 2 particular 3 federal
intervalos=[0,1.5,2.5,3.5]
nomes=['Pública','Particular','Federal']
aluno['Proveniência']=pd.cut(aluno['Proveniência'],bins=intervalos,labels=nomes)
aluno.head(3)

```

```python
# criar uma coluna nova
aluno['Média']=round((aluno['Nota1']+aluno['Nota2'])/2,2)
aluno.head(3)
```

```python
aluno.to_csv('/content/drive/MyDrive/Colab Notebooks/Alunos_corrigido.csv',encoding='utf-8',header=True,index=False)
```
